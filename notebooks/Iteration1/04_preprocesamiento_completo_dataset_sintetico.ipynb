## PREPROCESAMIENTO COMPLETO DEL DATASET SINTETICO

Generacion de dataset sintetico para prueba de limpieza

```python

import sys; sys.path.append('../..')

from src.utils.synthetic_generator import SyntheticDataGenerator
from src.utils.constants import SYNTHETIC_DATA_PATH

print("=" * 70)
print("GENERACIÓN DE DATASET DISRUPTIVO PARA PRUEBAS")
print("=" * 70)

# Crear generador con problemas extremos
gen_extreme = SyntheticDataGenerator(n_samples=400, seed=123)

gen_extreme.generate_clean_data()
gen_extreme.inject_missing_codes(percentage=0.15)
gen_extreme.inject_out_of_range_values({
    'q2': (0.10, 250),
    'edre': (0.08, 99),
    'q3cn': (0.12, 999)
})
gen_extreme.inject_true_nans(percentage=0.20)
gen_extreme.inject_outliers({'q12cn': (0.10, [20, 25, 30, 35])})

# Mostrar resumen
gen_extreme.print_summary()

# Guardar
synthetic_path = f"{SYNTHETIC_DATA_PATH}/problematic_test.dta"
gen_extreme.save_dataset(str(synthetic_path))

print(f"\n✓ Dataset extremo generado y listo para usar")
print("=" * 70)
```

Pipeline completo

```python
from src.preprocessing.preprocessor import DataPreprocessor
from src.utils.constants import RESULTS_PATH

print("=" * 70)
print("TEST PREPROCESSOR - DATASET SINTÉTICO")
print("=" * 70)

preprocessor = DataPreprocessor()

# Ejecutar pipeline completo con el dataset generado
print(f"\nEjecutando pipeline...")
df_final = preprocessor.run_pipeline(str(synthetic_path))

# Mostrar resultados
print(f"\n✓ Pipeline completado")
print(f"  Dimensiones: {df_final.shape}")
print(f"  Missings: {df_final.isnull().sum().sum()}")
print(f"  Códigos especiales: {(df_final.isin([888888, 988888, 999999])).sum().sum()}")

print(f"\nPrimeras 5 filas:")
print(df_final.head())

# Generar reporte
output_path = f"{RESULTS_PATH}/preprocessing/reporte_sintetico.json"
preprocessor.generate_report(str(output_path))
print(f"\n✓ Reporte guardado en: {output_path}")

print("\n" + "=" * 70)
```
